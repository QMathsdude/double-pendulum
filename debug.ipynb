```python
import tkinter as tk
from tkinter import font

# Create a root window (we don't need to show it)
root = tk.Tk()
root.withdraw()  # hide the main window

# Get all available font families
available_fonts = list(font.families())
available_fonts.sort()

# Print them
for f in available_fonts:
    print(f)

# Optional: show a simple GUI to select a font
def show_font(font_name):
    label.config(text=f"Sample Text in {font_name}", font=(font_name, 16))

root.deiconify()  # show window
root.title("Font Checker")

label = tk.Label(root, text="Sample Text", font=("Arial", 16))
label.pack(pady=10)

font_var = tk.StringVar(value="Arial")
font_menu = tk.OptionMenu(root, font_var, *available_fonts, command=show_font)
font_menu.pack(pady=10)

root.mainloop()
```
